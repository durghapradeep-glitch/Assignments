### Hypothesis testing

```python
### we are testing whether the actual weekly operating cost is higher than what the model predicts
```

```python
### The theoretical weekly operating cost model
### W=1000+5X
### X=number of units produced in a week
```

```python
mean_x=600
std_dev=25
```

```python
### theoretical mean weekly cost according to the cost mode
w1=1000+5*mean_x
w1
```

```python
### (H0):the actual weekly cost are not higher than the model suggests
### (H1):the actual weekly cost are higher than the model suggests
```

```python
sample_mean=3050
n=25 ### sample size=n
std_dev=25 
std_dev_w=5*std_dev ### std_dev_w=standard deviation of weekly cost
std_dev_w
```

```python
alpha=0.05
```

```python
# Compute standard error
SE = std_dev_w/np.sqrt(n)
```

```python
SE
```

```python
# Compute test statistic
z = (sample_mean - w1) / SE
```

```python
z
```

```python
### critical value
z_critical=stats.norm.ppf(1-alpha)
z_critical
```

```python
# Decision
if z > z_critical:
    print("Reject H0:the Actual weekly cost are higher than the model suggests")
else:
    print("Fail to reject H0:the actual weekly cost are not higher than the model suggests")
```
