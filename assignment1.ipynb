##### Exercise 1:Prime Numbers
###### Write a Python program that checks whether a given number is prime or not. A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself.

```python
import math
import random
```

```python
a=int(input('enter a value'))
if a<=1:
    print('invalid')
else:
    is_prime=True
    for i in range(2,int(a**0.5)+1):
      if (a%i)==0:
        print(a,' is not a prime number')
        is_prime=False
        break
if is_prime:
    print(a,' is a prime number')
```

### Exercise 4
##### Problem Statement (Question): Hospital Billing System

```python
room_charge=2000
doctor_consultation=1500
lab_test=300
days=int(input('how many days stayed'))
tests=int(input('how many lab tests taken'))
med=int(input('what is the medicine charge'))
the_total_bill=(days*room_charge)+doctor_consultation+(tests*lab_test)+med
if the_total_bill>10000:
    print('discount=',the_total_bill*(10/100))
    print('the final amount is',the_total_bill-the_total_bill*(10/100))
else:
    print('the final amount is',the_total_bill)
```
