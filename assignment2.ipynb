#### 11. Create a small DataFrame manually with 10 rows and columns: 'Name', 'Age', 'City', and 'Salary'. Then: - Use .info(), .describe() - Select 'Name' and 'City' columns - Drop 'City' column - Fill any missing values in 'Salary' column with the mean - Remove any duplicate row

```python
### creating a dataframe
data={'Name': ['Sandhya', 'Meivizhi', 'Nabeeha', 'Panda', 'Samson','Sanju','Durgha','Nandana','Nikhil','Yadhu'],
 'Age': [21, 23, 24, 22, 22,30,20,22,25,24],
 'City': ['bengaluru', 'pune', 'chennai', 'bengaluru', 'mumbai','delhi','kochi','hyderabad','jaipur','mumbai'],
 'Salary': [50000,80000,30000,np.nan,40000,50000,70000,np.nan,30000,50000]}
df_data=pd.DataFrame(data)
df_data
```

```python
## get information
df_data.info()
```

```python
df_data.describe()
```

```python
df_data[['Name','City']]
```

```python
### removing city from dataframe
df_data=df_data.drop('City',axis=1)
df_data
```

```python
### missing values
df_data.isnull().sum()
```

```python
### fill missing values in salary with mean
df_data['Salary']=df_data['Salary'].fillna(df_data['Salary'].mean())
```

```python
df_data
```

```python
df_data=df_data.drop_duplicates()
## the final data frame
df_data
```
